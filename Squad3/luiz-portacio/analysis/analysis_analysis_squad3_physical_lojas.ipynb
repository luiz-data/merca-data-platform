```python
# Carregar dados do banco
df_lojas = consultar_tabela("physical_lojas")
print(f"✅ physical_lojas carregado! Shape: {df_lojas.shape}")
```

```python
# Visão Geral
print("=" * 50)
print("🏪 VISÃO GERAL — physical_lojas")
print("=" * 50)

print(f"\n📐 Shape: {df_lojas.shape}")
print(f"   {df_lojas.shape[0]} lojas | {df_lojas.shape[1]} colunas")

print("\n📋 Colunas:")
print(df_lojas.dtypes)

df_lojas.describe()
```

```python
# Distribuição por Estado
print("=" * 50)
print("📊 DISTRIBUIÇÃO POR ESTADO")
print("=" * 50)

dist_estado = df_lojas['estado_loja'].value_counts().reset_index()
dist_estado.columns = ['estado', 'qtd_lojas']
print(dist_estado.to_string(index=False))
```

```python
# Distribuição por Cidade
print("=" * 50)
print("📊 DISTRIBUIÇÃO POR CIDADE")
print("=" * 50)

dist_cidade = df_lojas['cidade_loja'].value_counts().reset_index()
dist_cidade.columns = ['cidade', 'qtd_lojas']
print(dist_cidade.to_string(index=False))
```

```python
# Análise de Peso de Vendas
print("=" * 50)
print("📊 ANÁLISE DE PESO DE VENDAS")
print("=" * 50)

print(f"\n   Total  : {df_lojas['peso_vendas'].sum()}")
print(f"   Média  : {df_lojas['peso_vendas'].mean():.2f}")
print(f"   Máximo : {df_lojas['peso_vendas'].max()}")
print(f"   Mínimo : {df_lojas['peso_vendas'].min()}")

print("\n🏆 Top 5 lojas por peso de vendas:")
top5 = df_lojas[['nome_loja', 'cidade_loja', 'estado_loja', 'peso_vendas']] \
    .sort_values('peso_vendas', ascending=False) \
    .head(5)
print(top5.to_string(index=False))

print("\n🔻 Bottom 5 lojas por peso de vendas:")
bottom5 = df_lojas[['nome_loja', 'cidade_loja', 'estado_loja', 'peso_vendas']] \
    .sort_values('peso_vendas', ascending=True) \
    .head(5)
print(bottom5.to_string(index=False))
```

```python
# Conclusões
print("=" * 50)
print("📋 CONCLUSÕES — physical_lojas")
print("=" * 50)

print(f"""
   Total de lojas    : {len(df_lojas)}
   Estados presentes : {df_lojas['estado_loja'].nunique()}
   Cidades presentes : {df_lojas['cidade_loja'].nunique()}
   Peso total vendas : {df_lojas['peso_vendas'].sum()}
   Loja maior peso   : {df_lojas.loc[df_lojas['peso_vendas'].idxmax(), 'nome_loja']}
   Loja menor peso   : {df_lojas.loc[df_lojas['peso_vendas'].idxmin(), 'nome_loja']}
""")
```
